# Requisitos funcionales

RF-01. El sistema debe mostrar un mensaje de bienvenida con una explicación breve del cifrado al iniciar la aplicación.

RF-02. El sistema debe mostrar un menú con las opciones cifrar, descifrar con clave, probar todas las claves y salir.

RF-03. El sistema debe mostrar un mensaje de error y volver a pedir la opción cuando el usuario escriba un valor que no corresponda a ninguna opción del menú.

RF-04. El sistema debe solicitar el mensaje que se va a procesar en las opciones cifrar, descifrar con clave y probar todas las claves.

RF-05. El sistema debe mostrar un mensaje de error y volver a pedir el mensaje cuando esté vacío o tenga solo espacios.

RF-06. El sistema debe solicitar la clave después del mensaje en las opciones cifrar y descifrar con clave.

RF-07. El sistema debe mostrar un mensaje de error y volver a pedir la clave cuando no sea un número entero entre 1 y 26.

RF-08. El sistema debe reemplazar las vocales con tilde por las mismas vocales sin tilde antes de cifrar o descifrar.

RF-09. El sistema debe cifrar el mensaje reemplazando cada letra por la que está tantas posiciones adelante en el alfabeto de 27 letras como indique la clave.

RF-10. El sistema debe descifrar el mensaje reemplazando cada letra por la que está tantas posiciones atrás en el alfabeto de 27 letras como indique la clave.

RF-11. El sistema debe continuar desde el inicio del alfabeto cuando un desplazamiento hacia adelante pase de la zeta, y desde el final cuando un desplazamiento hacia atrás pase de la a.

RF-12. El sistema debe conservar las mayúsculas y las minúsculas de cada letra después de desplazarla.

RF-13. El sistema debe dejar sin cambios los espacios, los números y los signos de puntuación del mensaje.

RF-14. El sistema debe mostrar el texto claro y el texto cifrado, uno debajo del otro, al terminar de cifrar o descifrar.

RF-15. El sistema debe descifrar el mensaje con cada clave de 1 a 26 y mostrar los 26 resultados, cada uno junto a su clave, en la opción probar todas las claves.

RF-16. El sistema debe volver al menú principal después de mostrar cada resultado.

RF-17. El sistema debe mostrar la cantidad de mensajes procesados en la sesión y luego cerrar la aplicación cuando el usuario elija salir.
